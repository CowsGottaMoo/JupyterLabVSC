## Exercise 1: Sensor Data Cleanup and Combination

**Topic:**  
Creating, reshaping, slicing, filtering, and concatenating NumPy arrays of different sizes.

**Goal:**  
Practice building NumPy arrays from raw data, reshaping them into a useful structure, selecting specific parts of the data, filtering values based on conditions, and combining arrays with different lengths.

**Task:**  
You are given temperature measurements from two different sensors. Sensor A recorded 12 values, while Sensor B recorded 8 values.

Write your own code to:

1. Create NumPy arrays for both sensors.
2. Reshape the Sensor A data into a `3 x 4` array.
3. Select only the second row of the reshaped Sensor A data.
4. Filter all Sensor A values above `22.0`.
5. Filter all Sensor B values below `20.0`.
6. Concatenate the filtered values from both sensors into one final array.
7. Print the shape and contents of each important intermediate result.

**Starter data:**

```python

sensor_a = [19.8, 20.1, 21.5, 22.3,
            18.9, 19.5, 23.1, 24.0,
            20.7, 21.2, 22.8, 23.5]

sensor_b = [18.5, 19.2, 20.0, 21.1,
            17.8, 18.9, 22.4, 23.0]
```

---

````python
sensor_a = [19.8, 20.1, 21.5, 22.3,
            18.9, 19.5, 23.1, 24.0,
            20.7, 21.2, 22.8, 23.5]

sensor_b = [18.5, 19.2, 20.0, 21.1,
            17.8, 18.9, 22.4, 23.0]
````

````python
sensor_a = np.array(sensor_a)
sensor_b = np.array(sensor_b)

print(f"Sensor A shape: {sensor_a.shape}")
print(sensor_a)
print()
print(f"Sensor B shape: {sensor_b.shape}")
print(sensor_b)
````

Sensor A contains 12 values, so it can be reshaped into a matrix with 3 rows and 4 columns.

````python
reshaped_a = sensor_a.reshape(3, 4)

print(f"Reshaped Sensor A shape: {reshaped_a.shape}")
print(reshaped_a)
````

Just the second row can either be selected with index `1` or sliced with `1:2`

````python
second_row_sensor_a = reshaped_a[1:2]

print(f"Second row shape: {second_row_sensor_a.shape}")
print(second_row_sensor_a)
````

Boolean filters are used to select all Sensor A values above `22.0` and all Sensor B values below `20.0`. The filtered arrays have different lengths, but they can still be concatenated because both are one-dimensional arrays.

````python
filtered_a = sensor_a[sensor_a > 22.0]
filtered_b = sensor_b[sensor_b < 20.0]
combined_filtered_values = np.concatenate([filtered_a, filtered_b])

print(f"Sensor A values above 22.0 shape: {filtered_a.shape}")
print(filtered_a)
print()
print(f"Sensor B values below 20.0 shape: {filtered_b.shape}")
print(filtered_b)
print()
print(f"Combined filtered values shape: {combined_filtered_values.shape}")
print(combined_filtered_values)
````

The final combined array contains the high Sensor A measurements and the low Sensor B measurements. This shows how NumPy can reshape structured data, extract specific rows and combine conditionally selected values.

## Exercise 2: Polynomial Model for Noisy Measurements

**Topic:**
Working with polynomial functions using `numpy.polynomial.polynomial`, including evaluating a polynomial and fitting a polynomial curve to noisy toy data.

**Goal:**
Practice representing a polynomial with coefficients, evaluating it at different x-values, and fitting a polynomial to noisy data using NumPy’s polynomial tools.

**Task:**
You are given toy measurement data that roughly follows a quadratic trend, but with some noise.

Write your own code to:

1. Import `polynomial` from `numpy.polynomial`.
2. Create an array of x-values.
3. Define a polynomial with the coefficients `[2.0, -0.5, 0.25]`.
4. Evaluate this polynomial for all x-values.
5. Create or use the given noisy y-values.
6. Fit a degree-2 polynomial to the noisy data.
7. Evaluate the fitted polynomial on the same x-values.
8. Compare the original polynomial values, the noisy values, and the fitted values.

The coefficient array `[2.0, -0.5, 0.25]` represents the polynomial:

$$f(x) = 2.0 - 0.5x + 0.25x^2$$

**Starter data:**

```python
x_values = np.array([-4, -3, -2, -1, 0, 1, 2, 3, 4])

true_coefficients = np.array([2.0, -0.5, 0.25])

noisy_y_values = np.array([8.3, 5.9, 4.1, 2.8, 2.2,
                           1.7, 2.1, 3.0, 4.4])
```

***

````python
x_values = np.array([-4, -3, -2, -1, 0, 1, 2, 3, 4])

true_coefficients = np.array([2.0, -0.5, 0.25])

noisy_y_values = np.array([8.3, 5.9, 4.1, 2.8, 2.2,
                           1.7, 2.1, 3.0, 4.4])
````

````python
y_values = poly.polyval(x_values, true_coefficients)

print("True polynomial coefficients:")
print(true_coefficients)
print()
print("Evaluated true y-values:")
print(y_values)
````

A 2-degree polynomial is fitted to the noisy toy data. Afterwards the fitted coefficient array is evaluated with `poly.polyval()` as well.

````python
fitted_coefficients = poly.polyfit(x_values, noisy_y_values, deg=2)
fitted_y_values = poly.polyval(x_values, fitted_coefficients)

print("Fitted polynomial coefficients:")
print(np.round(fitted_coefficients, 4))
print()
print("Evaluated fitted y-values:")
print(np.round(fitted_y_values, 4))
````

The comparison table combines the x-values, the original polynomial values, the noisy values and the fitted values. This makes the differences easier to inspect numerically.

````python
comparison_table = np.column_stack((x_values, y_values, noisy_y_values, fitted_y_values))

print("Columns: x, true polynomial, noisy data, fitted polynomial")
print(np.round(comparison_table, 3))
````

The plot visualizes the same comparison: 
- The noisy measurements are shown as **individual points**,
- the original polynomial and fitted polynomial are shown as **lines**.
    * The original polynomial is displayed as a dashed line
    * The fitted polynomial is displayed as a solid line

````python
plt.plot()
plt.plot(x_values, noisy_y_values, 'o', markersize=15)
plt.plot(x_values, y_values, '-', linewidth=4)
plt.plot(x_values, fitted_y_values, '--', linewidth=3)

plt.xlabel('x-values')
plt.ylabel('y-values')
plt.title('True Polynomial Compared to Noisy and Fitted Values')

plt.grid()
plt.show()
````

The fitted polynomial follows the same quadratic trend as the original polynomial. It does not pass through every noisy point exactly, because the fit balances all measurement deviations to find a smooth degree-2 approximation.
