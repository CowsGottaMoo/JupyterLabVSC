## Solving the Tasks

#### 1: Creating 1D and 2D arrays

````python
# 1D array
arr1 = np.array([1, 2, 3, 4, 5])

# 2D array with shape (2, 3)
arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])

print("1D array:", arr1)
print("2D array:\n", arr2)
````

#### 2: Basic NumPy Functions

````python
sum_arr1 = np.sum(arr1)
mean_arr1 = np.mean(arr1)
max_arr2 = np.max(arr2)

print("Sum of 1D array:", sum_arr1)
print("Mean of 1D array:", mean_arr1)
print("Maximum value in 2D array:", max_arr2)
````

#### 3: Array Operations

````python
add_result = arr1 + 2
sub_result = arr1 - 1
mul_result = arr1 * 3
div_result = arr1 / 2

print("Addition:", add_result)
print("Subtraction:", sub_result)
print("Multiplication:", mul_result)
print("Division:", div_result)
````

#### 4: Create Arrays of Different Sizes

````python
array_a = np.array([[1, 2],
                    [3, 4]])

array_b = np.array([[5, 6, 7],
                    [8, 9, 10]])

print("Array A:\n", array_a)
print("Shape of A:", array_a.shape)

print("Array B:\n", array_b)
print("Shape of B:", array_b.shape)
````

#### 5: Concatenate Arrays

````python
arr1_extra = np.array([6, 7])
concat_1d = np.concatenate((arr1, arr1_extra))

print("Concatenated 1D array:", concat_1d)
````

````python
array_c = np.array([[11, 12],
                    [13, 14]])

concat_2d_rows = np.concatenate((array_a, array_c), axis=0)

print("Concatenated 2D array along rows:\n", concat_2d_rows)
````

#### 6: Handle Errors for Incompatible Concatenation

````python
try:
    # These arrays cannot be concatenated along axis=0 because their column counts differ
    bad_concat = np.concatenate((array_a, array_b), axis=0)
    print(bad_concat)
except ValueError as e:
    print("Concatenation failed:", e)
````

#### 7: Reflection Questions

1. Adding arrays of different shapes may work only if NumPy can broadcast them together. Otherwise, it raises an error.

2. `np.concatenate()` joins arrays together along an axis. Element-wise operations like addition or multiplication act on matching elements instead of joining arrays.

3. Arrays concatenate successfully only when their dimensions match in all axes except the one being concatenated.

#### Bonus Challenge

````python
def safe_concatenate(array1, array2, axis=0):
    try:
        result = np.concatenate((array1, array2), axis=axis)
        return result
    except ValueError as e:
        return f"Concatenation failed: {e}"
````

````python
# Works
print(safe_concatenate(arr1, arr1_extra))

# Works
print(safe_concatenate(array_a, array_c, axis=0))

# Fails
print(safe_concatenate(array_a, array_b, axis=0))
````
