## Exercise 1: Array Manipulation & Vectorized Speed Test
### Part A — Shape & filtering

1. Create a 1D NumPy array of integers from 0 to 23 (inclusive of 0, exclusive of 24).
2. Reshape it into a 4×6 array.
3. Slice out the second and third rows, all columns.
4. From the original 4×6 array, use boolean indexing to extract all values that are divisible by 3.
5. Concatenate the original 4×6 array with a new 2×6 array of your choice (values up to you), stacking them vertically so you end up with a 6×6 array.


1. Create a NumPy array of integers from 0 to 23.

```python
array = np.arange(24)
print(f'NumPy array of integers from 0 to 23: \n {array}')
```

2. Reshape the array to a 4x6 shape.

```python
array_reshaped = array.reshape(4, 6)
print(f'4x6 array: \n {array_reshaped}')
```

3. Slice out the second and third rows.

```python
second_row = array_reshaped[1:2]
third_row = array_reshaped[2:3]
print(f'The second row: {second_row} \nand the third row: {third_row}')
```

4. Extract values that are divisible by 3 using boolean indexing.

```python
filtered_items = np.where(array_reshaped % 3 == 0)
print(f'All values that are divisible by 3: \n {array_reshaped[filtered_items]}')
```

5. Concatenate the array with another array.

```python
new_array = np.array([np.arange(-12, -6), np.arange(-6, 0)])
array_long = np.concatenate([new_array, array_reshaped], axis=0)
print(f'After concatenating the two arrays, we get the following array: \n {array_long}')
```

### Part B — Loop vs. vectorized speed comparison

1. Create a large 1D array (e.g., 1,000,000 random floats).
2. Write a plain Python for loop that computes the square of each element and stores results in a list (or pre-allocated array).
3. Do the same computation using a vectorized NumPy operation.
4. Time both approaches (e.g., using time.time() or the timeit module) and print the speedup factor.
5. Bonus: try the same comparison with a slightly more complex operation, like sin(x) + x**2, and see if the speedup ratio changes.

1. Create a large array of random floats.

```python
large_array = np.random.random_sample(1000000)
```

2. and 3. Compute the square of each element with both a for loop and a NumPy operation.

```python
def loop_version(array: np.ndarray) -> list[float]:
    """ Square each element of a NumPy array using an explicit Python loop.
        
        Args:
            array: A NumPy array filled with values

        Return:
            values_squared: A list containing the squared values
    """
    
    values_squared = []
    for value in array:
        values_squared.append(value**2)
    return values_squared
        

def vectorized_version(array: np.ndarray) -> np.ndarray:
    """ Square each element of a NumPy array using a vectorized NumPy operation.
        
        Args:
            array: A NumPy array filled with values

        Return:
            values_squared: A NumPy array containing the squared values
    """
    values_squared = large_array**2
    return values_squared
```

4. Time boath approaches and compare the times.

```python
timeit.timeit(lambda:loop_version(large_array), number=20)
```

```python
timeit.timeit(lambda:vectorized_version(large_array), number=20)
```

As we can see, the vectorized NumPy operation is much faster.

## Exercise 2: Polynomial Fitting on Noisy Data

1. Using numpy.polynomial.polynomial (or the Polynomial class), define a polynomial of your choice — e.g., degree 3, with coefficients you pick.
2. Generate 50–100 x-values over some range (e.g., -5 to 5) using np.linspace.
3. Evaluate your polynomial at those x-values to get clean y-values.
4. Add Gaussian noise to the y-values (using np.random.normal) to simulate noisy measured data.
5. Use Polynomial.fit to fit a polynomial back to the noisy data. Try fitting with the correct degree, and then try with a degree that's too low and too high.
6. Print out the fitted coefficients vs. the true coefficients.

1. Define the polynomial.

```python
coefficients = [-4, 1, -2]
polynomial = poly.Polynomial(coefficients)  # coefficients = 3 -> degree = 2
```

2. Generate random x-values.

```python
x_values = np.linspace(-5, 5, num=80)
```

3. Compute the y-values with the Polynomial.

```python
y_values = polynomial(x_values)
```

4. Add noise to the y-values.

```python
noise = np.random.normal(size=len(y_values))
y_values_noisy = y_values + noise
```

5. Fit the polynomial back to the noisy data using a degree of 2 (as in the original Polynomial).

```python
poly_obj = poly.Polynomial.fit(x_values, y_values_noisy, deg=2)
```

6. Compare the fitted and true coefficients.

```python
fitted_coefficients = poly_obj.convert().coef
print(f'original coefficients: {coefficients} \nfitted coefficients: {fitted_coefficients}')
```

By comparing the coefficients, we can see that they are similar. Therefore, the fitting of the noisy data appears to be successful.
