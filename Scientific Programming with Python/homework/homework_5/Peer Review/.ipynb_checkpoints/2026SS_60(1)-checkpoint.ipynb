## Libraries

First, the allowed libraries that I will be needing are imported. NumPy is used for the numerical work [1, 3] and `timeit` is used for the small speed comparison. [4]

```python
# Importing the allowed libraries
import numpy as np
from numpy.polynomial import Polynomial
from numpy.polynomial import polynomial as polynomial_module

import timeit

# Some pre-settings I have read are good for this kind of work/exercise
np.set_printoptions(precision=4, suppress=True)
random_generator = np.random.default_rng(seed=60)
```

## Exercise 1: Array Operations

The first exercise reviews basic NumPy array work. The starting array is reshaped into a two-dimensional matrix, then different slices and filtered values are calculated from it. The row-centering step uses broadcasting. [3, 5]

```python
# Creating and reshaping the NumPy array
base_array = np.arange(1, 25)
number_matrix = base_array.reshape(4, 6)

# Extracting slices from the matrix
first_three_columns = number_matrix[:, :3]
last_two_rows = number_matrix[-2:, :]

# Filtering and summarizing the matrix
filtered_values = number_matrix[(number_matrix % 2 == 0) & (number_matrix > 10)]
row_means = number_matrix.mean(axis=1)
column_sums = number_matrix.sum(axis=0)

# Centering each row by subtracting the row mean
row_centered_matrix = number_matrix - row_means[:, np.newaxis]
combined_matrix = np.concatenate((number_matrix, row_centered_matrix), axis=1)

print("Original 4 x 6 matrix:")
print(number_matrix)

print("\nFirst three columns:")
print(first_three_columns)

print("\nLast two rows:")
print(last_two_rows)

print("\nEven values greater than 10:")
print(filtered_values)

print("\nRow means:")
print(row_means)

print("\nColumn sums:")
print(column_sums)

print("\nRow-centered matrix:")
print(row_centered_matrix)

print("\nCombined matrix shape:")
print(combined_matrix.shape)
```

## Exercise 2: Polynomial Fitting and Performance

The second exercise uses a quadratic polynomial. The coefficients are stored in increasing order of power, which matches the convention used by `numpy.polynomial.polynomial`. Two helper functions are defined so that the loop-based calculation and the vectorized calculation can be compared directly. [6]

```python
# Defining helper functions for polynomial evaluation
def evaluate_polynomial_with_loop(x_values: np.ndarray, coefficients: np.ndarray) -> np.ndarray:
    """Evaluate a polynomial with a standard Python loop.
    
    Args:
        x_values: NumPy array of x-values.
        coefficients: Polynomial coefficients in increasing order of power.
    
    Returns:
        NumPy array of calculated y-values.
    """
    if not isinstance(x_values, np.ndarray):
        raise TypeError("x_values must be a NumPy array")
    if not isinstance(coefficients, np.ndarray):
        raise TypeError("coefficients must be a NumPy array")
    
    y_values = []
    
    for x_value in x_values:
        y_value = 0.0
        
        for power, coefficient in enumerate(coefficients):
            y_value = y_value + coefficient * x_value ** power
        
        y_values.append(y_value)
    
    return np.array(y_values)


def evaluate_polynomial_vectorized(x_values: np.ndarray, coefficients: np.ndarray) -> np.ndarray:
    """Evaluate a polynomial with vectorized NumPy code.
    
    Args:
        x_values: NumPy array of x-values.
        coefficients: Polynomial coefficients in increasing order of power.
    
    Returns:
        NumPy array of calculated y-values.
    """
    if not isinstance(x_values, np.ndarray):
        raise TypeError("x_values must be a NumPy array")
    if not isinstance(coefficients, np.ndarray):
        raise TypeError("coefficients must be a NumPy array")
    
    return polynomial_module.polyval(x_values, coefficients)
```

The next cell generates noisy toy data from a known quadratic polynomial. We then fit a degree-2 polynomial to the noisy data and compare it to the original polynomial. In short we are trying to solve a very simple machine learning problem. [6]

```python
# Generating noisy toy data from a known quadratic polynomial
true_coefficients = np.array([1.5, -2.0, 0.3])
x_data = np.linspace(-5.0, 5.0, 31)

exact_y_data = evaluate_polynomial_vectorized(
    x_values=x_data,
    coefficients=true_coefficients
)
noise = random_generator.normal(loc=0.0, scale=0.8, size=x_data.shape)
noisy_y_data = exact_y_data + noise

# Fitting a degree-2 polynomial to the noisy toy data
fitted_polynomial = Polynomial.fit(x=x_data, y=noisy_y_data, deg=2).convert()
fitted_coefficients = fitted_polynomial.coef
fitted_y_data = fitted_polynomial(x_data)

root_mean_squared_error = np.sqrt(np.mean((fitted_y_data - exact_y_data) ** 2))

print("True coefficients:")
print(true_coefficients)

print("\nFitted coefficients:")
print(fitted_coefficients)

print("\nRoot-mean-squared error compared to the exact polynomial:")
print(f"{root_mean_squared_error:.4f}")

print("\nFirst five generated data points:")
print("x-value   noisy y-value   fitted y-value")

for row_index in range(5):
    print(f"{x_data[row_index]:7.3f}   {noisy_y_data[row_index]:13.4f}   {fitted_y_data[row_index]:14.4f}")
```

After all this processing and the calculations we compare the 2, namely the loop-based and the vectorized polynomial evaluations, using `timeit`. Of course there is a short results assertion to make sure they are what we expect them to be. [4]

```python
# Comparing loop-based and vectorized polynomial evaluation
large_x_values = np.linspace(-5.0, 5.0, 100_000)

loop_y_values = evaluate_polynomial_with_loop(
    x_values=large_x_values,
    coefficients=true_coefficients
)
vectorized_y_values = evaluate_polynomial_vectorized(
    x_values=large_x_values,
    coefficients=true_coefficients
)

same_results = np.allclose(loop_y_values, vectorized_y_values)

loop_time = timeit.timeit(
    lambda: evaluate_polynomial_with_loop(
        x_values=large_x_values,
        coefficients=true_coefficients
    ),
    number=3
)
vectorized_time = timeit.timeit(
    lambda: evaluate_polynomial_vectorized(
        x_values=large_x_values,
        coefficients=true_coefficients
    ),
    number=3
)
speedup_factor = loop_time / vectorized_time

print(f"The two methods give the same results: {same_results}")
print(f"Loop time for 3 runs:       {loop_time:.4f} seconds")
print(f"Vectorized time for 3 runs: {vectorized_time:.4f} seconds")
print(f"Vectorized speedup factor:  {speedup_factor:.1f} times faster")
```

## Comment on the Results

The first exercise shows how a single NumPy array can be reshaped, sliced, filtered, summarized, and combined with another array. Broadcasting is useful in the row-centering step because each row mean can be subtracted from the corresponding row without writing a manual nested loop. [3, 5]

The second exercise shows that `Polynomial.fit` can recover a quadratic polynomial from noisy toy data. The fitted coefficients are close to the true coefficients, but not exactly identical because random noise was added to the data. The `timeit` comparison also shows the normal scientific programming reason for using NumPy: the vectorized calculation gives the same numerical result as the loop-based calculation, but it runs **much** faster.
