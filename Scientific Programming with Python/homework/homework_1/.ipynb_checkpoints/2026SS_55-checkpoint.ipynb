In the following I am going to calculate the energy that is released during the beta decay of a single neutron using Einstein's equation E = mc^2 where 'E' is energy, 'm' is the mass weight and 'c' is the speed of light in m/s. 

```python
mass_neutron = 1.67492749804e-27
mass_proton =  1.67262192369e-27
mass_electron = 9.1093837015e-31

c = 299792458

m = mass_neutron - (mass_proton + mass_electron)

E = m * (c ** 2)

print("The energy that is released by the beta decay of a single neutron within an atom is: ", E , "J" )
```
