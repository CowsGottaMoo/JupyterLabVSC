# Assignment: Mass to Energy - Beta Decay


## Introduction

In beta decay, a neutron is converted into a proton, an electron, and released energy.

The released energy is caused by the mass difference between the neutron and the sum of the proton and electron masses.

To calculate the released energy, Einstein's equation is used:

\[
E = mc^2
\]

where $E$ is the energy, $m$ is the mass difference, and $c$ is the speed of light.


```python
# masses in kg
mass_neutron = 1.67492749804e-27
mass_proton = 1.67262192369e-27
mass_electron = 9.1093837015e-31

# speed of light in m/s
speed_of_light = 299792458

mass_difference = mass_neutron - (mass_proton + mass_electron)
released_energy = mass_difference * speed_of_light**2

print(f"Mass difference: {mass_difference:.6e} kg")
print(f"Released energy: {released_energy:.6e} J")
```

## Interpretation

The result shows that the neutron has slightly more mass than the combined masses of the proton and electron.

This small missing mass is converted into energy during beta decay.

For one beta decay event, the released energy is approximately $1.25 \\times 10^{-13}$ J.
