```python
#SciPro_ID = 2026SS_07
#import this

neutron_m = 1.67492749804e-27
proton_m = 1.67262192369e-27
electron_m = 9.1093837015e-31
speed_of_light = 299792458

delta = neutron_m - (proton_m + electron_m) # neutron -> proton + electron + energy therefore: energy = neutron - (proton + electron)

energy = delta * speed_of_light**2 #E = m*c**2

print("Mass Difference: ", delta)
print("Actual Energy: ", energy)
```
