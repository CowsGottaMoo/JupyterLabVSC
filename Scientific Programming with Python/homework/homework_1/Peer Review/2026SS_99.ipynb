## Task 1: Energy from Beta Decay

During beta decay, a neutron decays into proton and an electron. The mass difference is converted into energy.

Formula: E=m*c^2

Where: m = mass difference c = speed of light (3.0 x 10^8 m/s)

```python

m_neutron = 1.67492749804e-27
m_proton = 1.67262192369e-27
m_electron = 9.1093837015e-31

c = 3.0e8

#massdiffrence 
m_diff =  m_neutron - (m_proton + m_electron)


E = m_diff * c**2

print("Massdifference: ", m_diff, "kg")
print("released energy: ",E , "joule")
print(f"energy in scientific notation: {E:.3e}J")
```
