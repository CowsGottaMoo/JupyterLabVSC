### Task 1

The goal of this task is to create a user-defined function that transcribes a DNA sequence into its RNA equivalent.

DNA is composed of the nucleotide bases adenine (`A`), thymine (`T`), cytosine (`C`), and guanine (`G`). During transcription, thymine is replaced by uracil (`U`).

For this we will use the following replacement:

$$T \rightarrow U$$

where:
- `T` is thymine in the original **DNA** sequence,
- `U` is uracil in the resulting **RNA** sequence.

The function first checks that the input is a string and then checks whether only valid DNA bases are used. The sequence is converted to uppercase so that lower-case input can be handled as well.

```python
def dna_transcription(dna_string: str) -> str:
    '''
    Transcribe a DNA sequence to its RNA equivalent.

    This function replaces every T (thymine) in a DNA sequence with U (uracil).
    The input is handled case-insensitively by converting the sequence to uppercase
    before transcription.

    Args:
        dna_string: A DNA sequence containing the bases A, T, C, and G.

    Returns:
        The RNA equivalent of the given DNA sequence as a string.

    Raises:
        TypeError: If dna_string is not a string.
        ValueError: If dna_string contains characters other than A, T, C, or G.
    '''
    if not isinstance(dna_string, str):
        raise TypeError(f'The value used for DNA: {dna_string} is not a valid String.')

    dna_string = dna_string.upper()

    if any(char not in {'A', 'T', 'C', 'G'} for char in dna_string):
        raise ValueError(
            f'The string used for DNA transcription contains invalid characters: {dna_string}. '
            f'Only A, T, C and G are allowed.'
        )

    return dna_string.replace('T', 'U')
```

The function returns an RNA string after replacing every `T` with `U`. Invalid input is rejected before the replacement step so that corrupted sequence data are not processed further.

***

### Task 2

The goal of this task is to use unit tests to verify the behavior of the transcription function.

The unit tests check the following cases:
- a standard DNA sequence is transcribed correctly,
- lower-case input is handled correctly,
- an invalid character raises a `ValueError`,
- a non-string input raises a `TypeError`.

The first three tests are required by the assignment. The additional type test checks the type validation that is implemented in the function.

```python
class UnitTest(unittest.TestCase):

    def test_standard_sequence(self):
        self.assertEqual('AUCGAU', dna_transcription('ATCGAT'))

    def test_case_insensitivity(self):
        self.assertEqual('AUCG', dna_transcription('atcg'))

    def test_invalid_char(self):
        with self.assertRaises(ValueError):
            dna_transcription('X')

    def test_invalid_type(self):
        with self.assertRaises(TypeError):
            dna_transcription(123)


test_suite = unittest.TestLoader().loadTestsFromTestCase(UnitTest)
test_result = unittest.TextTestRunner(verbosity=2).run(test_suite)
```

If all tests return `OK`, the transcription function behaves correctly for the tested standard case, lower-case input, invalid characters, and invalid input types.

### Task 3

The goal of this task is to create toy data by randomly generating one DNA sequence.

The generated sequence should only contain the DNA bases `A`, `T`, `C`, and `G`. Its length is also assigned randomly.

For this we will use the following range:

$$10 \leq n \leq 100$$

where:
- `n` is the number of characters in the generated DNA sequence,
- each character is randomly selected from `A`, `T`, `C`, and `G`.

This kind of artificial sequence can be used as simple test data for future sequence-processing code.

```python
def toy_data(chars: str = 'ATCG') -> str:
    '''
    Randomly generate one DNA sequence string.

    The sequence is created by randomly selecting characters from the valid DNA
    bases A, T, C, and G. The length of the sequence is randomly assigned between
    10 and 100 characters.

    Args:
        chars: The characters from which the DNA sequence is generated.

    Returns:
        A randomly generated DNA sequence as a string.
    '''
    size = random.randint(10, 100)
    data = ''.join(random.choice(chars) for _ in range(size))

    print(f'Randomly generated DNA sequence: {data} containing {size} characters.')
    return data
```

The function is called once below to generate an example DNA sequence. Since the sequence is random, the result can be different each time the cell is executed.

```python
generated_sequence = toy_data()
generated_sequence
```
