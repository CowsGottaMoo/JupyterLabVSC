### Task 1

The goal of this task is to create two user-defined functions. 

The first function calculates the distance an object falls under the standard acceleration of gravity while ignoring air resistance.

For this we will use following equation:
   $$d = \frac{1}{2} \cdot g \cdot t^2$$
   where:
- `d` is the distance fallen in **meters**,
- `g` is the standard acceleration of gravity in **meters per second squared**,
- `t` is the time spent falling in **seconds**.

```python
def distance_falling(gravity, time):
    distance = 0.5 * gravity * time**2
    print(f"After {time} seconds, the object has fallen {distance:.2f} meters under a constant acceleration of {gravity} m/s².")
    return distance
```

The second function calculates the average speed of a falling object over a given time.

For this we will use following equation:
   $$v_{avg} = \frac{d}{t}$$
   where:
   - `v_avg` is the average speed in **meters per second**,
   - `d` is the distance fallen in **meters**,
   - `t` is the time spent falling in **seconds**.

```python
def average_speed(distance, time):
    avg_speed = distance / time
    print(f"The average speed of an object that fell {distance:.2f} meters in {time} seconds is {avg_speed:.2f} m/s.")
    print()
    return avg_speed
```

### Task 2
The goal of this task is to use the above functions to compute the average speed of a falling object under different gravitational accelerations and for different falling times.

The gravitational accelerations used are as follows:
- Earth: $g$ = 9.80665 $m/s$ $^{[1]}$
- Moon: $g$ = 1.62468 $m/s$ $^{[2]}$
- Mars: $g$ = 3.70875 $m/s$ $^{[3]}$

The average speed will be calculated for an object that falls for the following durations:
- 10.0 seconds
- 20.0 seconds
- 30.0 seconds
- 40.0 seconds
- 50.0 seconds

```python
earth_g = 9.80665
moon_g = 1.62468
mars_g = 3.70875

durations = [10.0, 20.0, 30.0, 40.0, 50.0]
```

#### Earth gravity

```python
for times in durations:
    distance = distance_falling(earth_g, times)
    average_speed(distance, times)
```

#### Moon gravity

```python
for times in durations:
    distance = distance_falling(moon_g, times)
    average_speed(distance, times)
```

#### Mars gravity

```python
for times in durations:
    distance = distance_falling(mars_g, times)
    average_speed(distance, times)
```

***

#### Summary of Results and Interpretation

The results show that the falling distance and the average speed both increase as the falling time increases.

The falling distance increases quadratically with time. This means that when the falling time doubles, the distance fallen becomes four times larger.

The average speed increases linearly with time. Resulting in the following: stronger gravitational acceleration leads to a larger average speed for the same falling time.

Comparing the three gravitational accelerations, Earth produces the largest falling distances and average speeds, the Moon produces the smallest values, and Mars lies between them. 
After 50 seconds, the object falls about 12258.31 meters on Earth, about 4635.94 meters on Mars, and about 2030.85 meters on the Moon.

These differences stems from the fact that the gravitational acceleration is strongest on Earth, weaker on Mars, and weakest on the Moon.
