***Goal***: To implement user-defined functions in Python to calculate the distance and average speed of a falling object under constant gravitational acceleration and to compare the results for Earth, the Moon, and Mars at different time intervals.

***Context***: The motion of a falling object is determined by the acceleration due to gravity [1], which depends on the celestial body (see Figure 1). Using known physical equations, the distance and average velocity can be calculated as functions of time. This allows for a comparison of the object's behavior under different gravitational conditions.



***Figure 1***: Acceleration due to gravity for different celestial bodies [2]

### Formulas

The distance traveled by a falling object under constant acceleration is given by:

$$ d = \frac{1}{2} \cdot g \cdot t^2  $$

where ***d*** is distance, ***g*** is the standard acceleration of gravity(see Figure 1) and ***t*** is the time that the object spent falling
 

The average velocity of the falling object can be calculated using:

$$ v = \frac{d}{t} $$


### Method

#### Task 1: Functions for formulas

```python
def distance(acceleration: float, time: float | int) -> float:
    """Computes distance of a falling object.

    Parameters:
    acceleration: gravitational acceleration (m/s^2)
    time: falling time (s)

    Returns:
    distance (m) """

    if not isinstance(acceleration, float):
        raise TypeError(f"The value for the acceleration ({acceleration}) must be a float type.")
    elif not isinstance(time, (float,int)):
        raise TypeError(f"The value for the time ({time}) must be a float or an int type.")
    if time < 0 or acceleration < 0:
        raise ValueError("acceleration and time can not be negative")
    else:
        return 1/2 * acceleration * time**2
```

```python
def avg_speed(distance: float, time: float | int) -> float:
    """Computes average velocity of a falling object.

    Parameters:
    distance: distance travelled (m)
    time: falling time (s)

    Returns:
    average velocity (m/s)
    """

    if not isinstance(distance, float):
        raise TypeError(f"The value for the distance ({distance}) must be a float type.")
    elif not isinstance(time, (float,int)):
        raise TypeError(f"The value for the time ({time}) must be a float or an int type.")
    if distance < 0 or time < 0:
        raise ValueError("acceleration and time can not be negative")
    else: 
        return distance/time
```

#### Task 2: Examples

In this task, the average velocity of a falling object is computed for Earth, Moon, and Mars using different time intervals (10 to 50 seconds). The results are calculated using the previously defined functions and compared.

```python
g_earth = 9.80665 # Acceleration on Earth (m/s^2)
g_moon = 1.62468 # Acceleration on Moon (m/s^2)
g_mars = 3.70875 # Acceleration on Mars (m/s^2)
```

```python
times = [10.0, 20.0, 30.0, 40.0 ,50.0] #time intervals
```

```python
planets = {
    "Earth": g_earth,
    "Moon": g_moon,
    "Mars": g_mars
}
```

```python
for planet, gravity in planets.items():
    print(f"\nThe average speed of the falling object on {planet} is:\n")

    for t in times:
        d = distance(gravity, t) #distance
        speed = avg_speed(d, t)
        print (f"t = {t}s -> v = {speed:.3e} m/s")      
```

### Conclusion 

The average velocity increases with time and is highest on Earth, lower on Mars, and lowest on the Moon due to different gravitational accelerations.
