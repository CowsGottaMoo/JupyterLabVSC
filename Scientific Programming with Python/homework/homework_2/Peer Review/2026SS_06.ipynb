**Task 1**: Create user-defined functions for computing the distance

(Equation 1) and the average speed (Equation 2) of a falling
         object as a function of the acceleration of gravity time

For that we first have to calculate the distance

```python
def calculate_distance(g: int|float, time: int|float) -> int|float:
  '''Calculates distance an object falls after a certain amount of time
     Args:
          g: The gravitys acceleration that is defined
          time: The time the object falls
     Return:
            returns the distance
  '''

  if not isinstance(g, int|float):
      raise TypeError("G is a defined number, please correct it")
  elif not isinstance(time, int|float):
      raise TypeError("You have to type in a number")
  else:
      distance = 0.5 * g * (time**2)
      return distance


```

If we have the distance for the falling object, we can calculate the speed.

```python
def calculate_average_speed(distance: int | float, time: int| float) -> float|int:
  '''Calculates the average speed for for a falling object
     Args:
          distance: The distance an object falled after a certain period of time
          time: The time the object falls
     Return:
            returns the average speed of a falliing object
  '''
  if not isinstance(distance, int|float):
      raise TypeError("You have to type in a number")
  elif not isinstance(time, int|float):
      raise TypeError("You have to type in a number")
  else:
      average_speed = distance / time
      return average_speed
```

So this is the way to calculate the different speeds

**Task 2**:  For the Earth, the Moon, and Mars, compute the average

speed of a falling object that falls for the following times:

10.0, 20.0, 30.0, 40.0, and 50.0 seconds. Comment on the results

For that, I would use a single method that can calculate the average speeds for different planets, because the logic behind the method is the same. Only the value of g changes.

```python
def calculate_average_speed_forPlanet(g: float | int ) -> None:
  '''Calculates the average speed for the Planet you want for different times
     Args:
          g: The gravitys acceleration that is defined
     Return:
            prints the average speed of a falliing object for different times
  '''
  if not isinstance(g,  int|float):
    raise TypeError("G is a defined number you have to type in")
  else:
     time = 10.0 # we start bei 10.0
  while time <= 50.0: # loop continues until time equals 50.0
    total_distance = calculate_distance(g, time) #call the methods
    total_speed = calculate_average_speed(total_distance, time)
    print(f"the average speed for {time} seconds is {total_speed} m/s")
    time += 10.0



```

You can use this method for all three planets, you just have to type in the g

This would be for the Earth

```python
calculate_average_speed_forPlanet(g = 9.80665)
```

This is for the Moon

```python
calculate_average_speed_forPlanet(g = 1.62468)
```

Last but no Least the Mars

```python
calculate_average_speed_forPlanet(g = 3.70875)
```

If you compare all three results, you can see that the average speed on Earth is the highest. On the Moon it is much slower and the average speed on the Mars is something between them. The stronger the gravity, the faster the average speed.
