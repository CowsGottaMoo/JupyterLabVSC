This notebook calculates how fast objects fall on Earth, the Moon, and Mars. I used the formulas for distance and average speed.

```python
def calculate_distance(gravity: float, time: float) -> float:
    """
    Calculates falling distance.
    Formula: d = 0.5 * g * t^2 [4]
    """

    if not isinstance(gravity, (int, float)):
        raise TypeError("Gravity must be a number.")
    elif not isinstance(time, (int, float)):
        raise TypeError("Time must be a number.")
    else:
        distance = 0.5 * gravity * (time ** 2)
        return distance


def calculate_average_speed(gravity: float, time: float) -> float:
    """
    Calculates average speed using the distance function.
    Formula: v_avg = d / t [4]
    """
    if not isinstance(gravity, (int, float)):
        raise TypeError("Gravity must be a number.")
    elif not isinstance(time, (int, float)):
        raise TypeError("Time must be a number.")
    elif time <= 0:
        raise ValueError("Time must be greater than zero.")
    else:
        dist = calculate_distance(gravity, time)
        return dist / time
```

```python
# Constants from the papers [1, 2, 3]
planets = {
    'Earth': 9.80665,
    'Moon': 1.62468,
    'Mars': 3.70875
}

times = [10.0, 20.0, 30.0, 40.0, 50.0]

# Looping through planets and times
for planet, g in planets.items():
    print(f"--- {planet} ---")
    for t in times:
        speed = calculate_average_speed(g, t)
        print(f"At {t}s, average speed is {speed:.2f} m/s")
    print() # Prints an empty line
```

## Comments
* Speed goes up linearly as time passes.
* Earth has the highest gravity (9.8 m/s^2), so the speeds are much higher here. The Moon has very low gravity, so objects fall much slower.
