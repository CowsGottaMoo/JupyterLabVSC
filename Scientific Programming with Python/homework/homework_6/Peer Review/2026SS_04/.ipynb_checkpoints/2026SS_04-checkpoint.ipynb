## 3. Imports

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

## 4. Load the CSV Dataset

The notebook reads the dataset from a relative CSV path. This makes the notebook easier for peers to run if the CSV file is kept in the same folder as the notebook.

```python
data_path = Path("study_habits_dataset.csv")

if not data_path.exists():
    raise FileNotFoundError(
        "The CSV file 'study_habits_dataset.csv' must be in the "
        "same folder as this notebook."
    )

df = pd.read_csv(data_path)

print(f"Dataset shape: {df.shape[0]} rows and {df.shape[1]} columns")
df.head()
```

## 5. Initial Data Checks

Before creating plots, the dataset is checked for missing values, data types, and basic numerical summaries. This separates data loading from data validation and makes the workflow easier to review.

```python
print("Column data types:")
print(df.dtypes)

print("\nMissing values per column:")
print(df.isna().sum())
```

```python
numeric_cols = [
    "study_hours_per_week",
    "sleep_hours_per_night",
    "attendance_percent",
    "practice_tests_completed",
    "exam_score_percent",
]

df[numeric_cols].describe().round(2)
```

## 6. User-Defined Helper Function

The helper function below provides a concise summary by course type. It includes type hints, a docstring, and internal input checks.

```python
def summarize_by_group(
    data: pd.DataFrame,
    group_col: str,
    value_col: str,
) -> pd.DataFrame:
    """
    Summarize a numeric column by a categorical group.

    Parameters
    ----------
    data : pd.DataFrame
        Input dataset containing the grouping and value columns.
    group_col : str
        Name of the categorical column used for grouping.
    value_col : str
        Name of the numeric column to summarize.

    Returns
    -------
    pd.DataFrame
        Summary table containing count, mean, median, and standard
        deviation for each group.

    Raises
    ------
    TypeError
        If data is not a Pandas DataFrame.
    ValueError
        If group_col or value_col is missing from the dataset.
    """
    if not isinstance(data, pd.DataFrame):
        raise TypeError("data must be a Pandas DataFrame.")
    elif group_col not in data.columns:
        raise ValueError(f"{group_col} is not a column in the dataset.")
    elif value_col not in data.columns:
        raise ValueError(f"{value_col} is not a column in the dataset.")

    summary = (
        data.groupby(group_col)[value_col]
        .agg(["count", "mean", "median", "std"])
        .round(2)
    )

    return summary


score_summary = summarize_by_group(
    data=df,
    group_col="course_type",
    value_col="exam_score_percent",
)

score_summary
```

## 7. Visualization 1: Distribution of Exam Scores

A histogram shows the distribution of exam scores. This helps answer whether most fictional students have low, medium, or high scores.

```python
fig, ax = plt.subplots(figsize=(8, 5))

sns.histplot(
    data=df,
    x="exam_score_percent",
    bins=12,
    kde=True,
    ax=ax,
)

ax.set_title("Distribution of Exam Scores")
ax.set_xlabel("Exam score (%)")
ax.set_ylabel("Number of students")

plt.tight_layout()
plt.show()
```

### Observation

The exam scores are mostly concentrated in the higher-middle range. Because the dataset is synthetic, this pattern reflects the chosen data-generation process rather than a real student population.

## 8. Visualization 2: Study Hours and Exam Scores

A scatter plot is useful for examining the relationship between two numerical variables. Here, study hours per week are compared with exam score.

```python
fig, ax = plt.subplots(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="study_hours_per_week",
    y="exam_score_percent",
    hue="course_type",
    ax=ax,
)

ax.set_title("Study Hours and Exam Score")
ax.set_xlabel("Study hours per week (hours)")
ax.set_ylabel("Exam score (%)")
ax.legend(title="Course type")

plt.tight_layout()
plt.show()
```

### Observation

The scatter plot suggests a positive relationship between study hours and exam score. Students with more weekly study hours often have higher exam scores in this synthetic dataset.

## 9. Visualization 3: Exam Scores by Course Type

A box plot compares exam score distributions across course types. This shows whether one course group appears to have higher or lower scores than another.

```python
fig, ax = plt.subplots(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="course_type",
    y="exam_score_percent",
    ax=ax,
)

ax.set_title("Exam Scores by Course Type")
ax.set_xlabel("Course type")
ax.set_ylabel("Exam score (%)")

plt.tight_layout()
plt.show()
```

### Observation

The box plot shows that the median exam scores are broadly similar across the three course types. Differences between groups are smaller than the overall spread within each group.

## 10. Visualization 4: Correlation Analysis

A correlation heatmap summarizes linear relationships between numerical variables. Correlation does not prove causation, but it is useful for exploring patterns in data [4].

```python
corr = df[numeric_cols].corr()

fig, ax = plt.subplots(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    linewidths=0.5,
    ax=ax,
)

ax.set_title("Correlation Matrix of Numerical Variables")

plt.tight_layout()
plt.show()
```

### Observation

The correlation matrix suggests that study hours, attendance, sleep, and completed practice tests are positively related to exam score in this synthetic dataset. The strongest relationship is expected to involve study hours because the data were generated so that study time contributes directly to the exam score.

## 11. Visualization 5: Trend Line with Polynomial Fit

A first-degree polynomial fit is added to show the overall trend between study hours and exam scores. This is a simple linear trend model and should be interpreted as exploratory, not predictive.

```python
x_vals = df["study_hours_per_week"].to_numpy()
y_vals = df["exam_score_percent"].to_numpy()

coef = np.polyfit(x_vals, y_vals, deg=1)
trend = np.poly1d(coef)

x_line = np.linspace(x_vals.min(), x_vals.max(), 100)
y_line = trend(x_line)

fig, ax = plt.subplots(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="study_hours_per_week",
    y="exam_score_percent",
    ax=ax,
)

ax.plot(x_line, y_line)

ax.set_title("Linear Trend Between Study Hours and Exam Score")
ax.set_xlabel("Study hours per week (hours)")
ax.set_ylabel("Exam score (%)")

plt.tight_layout()
plt.show()

print(f"Fitted line: exam_score = {coef[0]:.2f} * study_hours + {coef[1]:.2f}")
```

### Observation

The fitted line slopes upward, which supports the visual observation that more weekly study hours are associated with higher exam scores in this dataset.
