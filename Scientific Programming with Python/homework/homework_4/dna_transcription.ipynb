## Task 1: Transcription Function

The goal of this task is to create a user-defined function that transcribes a DNA sequence into its RNA equivalent.$^{[1]}$

DNA is composed of four nucleotide bases:$^{[2]}$
- `A` (adenine)
- `T` (thymine)
- `C` (cytosine)
- `G` (guanine)
  
During transcription, thymine is replaced by uracil `U`.

The function first checks that the input is a string and then checks whether only valid DNA bases are used. The sequence is converted to uppercase so that lower-case input can be handled as well.

```python
def dna_transcription(dna_string: str) -> str:
    ''' Transcribes a DNA sequence to its RNA equivalent.
        This Function replaces every T (thymine) with U (uracil) to transcribe DNA to RNA.

        Args: 
            dna_string: A DNA sequence containing the bases A, T, C, and G.
            
        Returns:
            The RNA equivalent to the given DNA sequence as a string.
            
        Raises:
            TypeError: If the dna_string (DNA sequence) is not a string.
            ValueError: If the dna_string contains characters other than: A, T, C, and G.
    '''
    if not isinstance (dna_string, str):
        raise TypeError(f'The value used for DNA: {dna_string} is not a valid String.')

    dna_string = dna_string.upper()

    if any(char not in {'A', 'T', 'C', 'G'} for char in dna_string):
        raise ValueError(
            f'The string used for DNA transcription contains invalid characters: {dna_string}. '
            f'Only A, T, C and G are allowed.')
            
    return(dna_string.replace('T', 'U'))
```

The function returns a RNA string after replacing every `T` with `U`. Invalid input is rejected before the replacement step so that corrupted sequence data are not processed further.

## Task 2: Unit Test
The goal of this task is to use unit tests to verify the behavior of the transcription function.

The unit tests check the following cases:
- a standard DNA sequence is transcribed correctly.
- lower case input is handled correctly.
- an invalid character raises a `ValueError`.
- a non string input raises a `TypeError`.

The first three tests are required by the assignement. The additional type test checks the type validation that is implemented in the function.

```python
class UnitTest(unittest.TestCase):

    def test_standard_sequence(self):
        self.assertEqual('AUCGAU', dna_transcription('ATCGAT'))

    def test_case_insensitivity(self):
        self.assertEqual('AUCG', dna_transcription('atcg'))

    def test_invalid_char(self):
        with self.assertRaises(ValueError):
            dna_transcription('X')

    def test_invalid_type(self):
        with self.assertRaises(TypeError):
            dna_transcription(123)

unittest.main(argv=[''], exit=False)
```

If all tests return `OK`, the transcription function behaves correctly for the tested standard case, lower case input, invalid characters and invalid input types.

## Task 3: Toy Data
The goal of this task is to create toy data by randomly generating one DNA sequence.

The generated sequence should only contain the DNA bases: `A`, `T`, `C` and `G`. Its length is also assigned randomly.For this we will use a range between 10 and 100 characters

The determining factor for the arrangment of the bases and the length of the sequence is the `random` library.

This kind of artificial sequence can be used as simple test data for future sequence processing code.$^{[3]}$

```python
def toy_data(chars='A' + 'T' + 'C' + 'G') -> str:
    '''Randomly generate one DNA sequence string.

       The sequence is created by randomly selecting characters from the valid DNA
       bases A, T, C, and G. The length of the sequence is randomly assigned between
       10 and 100 characters.

       Args:
           chars: The characters from which the DNA sequence is generated.
       
       Returns:
           A randomly generated DNA sequence as a string.
    ''' 
    size = random.randint(10, 100)
    data = ''.join(random.choice(chars) for _ in range(size))
    
    #print(f'Randomly generated DNA sequence: {data} containing {size} characters.')
    return data
```

The function is called once below to generate an example DNA sequence. Since the sequence is random, the result can be diffrent each time the cell is executed.

```python
gen_data = toy_data()
print(gen_data)
```
