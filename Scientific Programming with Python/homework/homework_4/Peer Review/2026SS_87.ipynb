## Task 1 DNA to RNA Transcription Function
this is a function that converts DNA pairs to its RNA equivalent (e.g., ATTGT becomes AUUGU)
for this we can use the str.replace() method

```python
def DNAtoRNA(dna) -> str:

    """ 
    Transcribes a DNA string to an RNA equivalent string by replacing (T)hymine with (U)racil
    --------------------------------------------------
    
    Parameters: 
    dna (str): DNA String e.g., "ATTGCTAGT"

    Returns:
    rna (str): RNA equivalent to a DNA string e.g, AUUGCUAGU
    """
    
    # Check if type is String
    if not isinstance(dna, str):
        raise ValueError("not a String")

    tmp = dna.upper() # Make sequence uppercase
    
    allowed = set("ACGT")
    # Check if String contains only allowed characters
    if not set(tmp).issubset(allowed):
        raise ValueError("Invalid character found. Sequence must only contain A, C, G or T")

    # Replace T with U
    rna = tmp.replace("T", "U")
    
    return rna
    
```

## Task 2 Unit Test
We want to make sure our code is up to a certain standard, for this we can use unit testing by using the unittest library
unit testing helps us to avoid certain errors 

Also unit testing helps with making sure future changes don't break the basic functionality of the code

```python

class TestDNAtoRNA(unittest.TestCase):
    # simple test to check if T is replaced by U in a given string
    def testReplace(self):
        self.assertEqual(DNAtoRNA("ATT"), "AUU")
        self.assertEqual(DNAtoRNA("AGTCGGGT"), "AGUCGGGU")

    # Testing invalid type overloading should raise an error
    def testTypeException(self):
        try:
            DNAtoRNA('X')
        except ValueError:
            pass
        else:
            raise AssertionError("Value Error never called")
    
    # Test invalid characters should raise an error
    def testInvalidCharacterException(self):
        try:
            DNAtoRNA("X")
        except ValueError:
            pass
        else:
            raise AssertionError("Invalid Character Exception not or wrongly implemented")
        
if __name__ == '__main__':
    unittest.main(argv=['ignored', '-v'], exit=False)
    
```

## Task 3 Generating Toy Data
The following function generates a single randomly generated DNA string composed of nucleotides A, T, C and G with a random length between 10 and 100 characters

```python
def generateToyData() -> str:
    bases = ["A","C", "T", "G"]
    baseLength = random.randrange(10, 101)
    dna = ""
    for x in range(baseLength):
        dna = dna+random.choice(bases)

    return dna
    
print (generateToyData())
    
```
