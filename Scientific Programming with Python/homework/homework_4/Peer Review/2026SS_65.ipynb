# Task 1: DNA to RNA Transcription Function
in this task I am going to create a function that transform a DNA string into its RNA form


```python
def transcribe_dna_to_rna(dna_sequence):
    """
    Transcribe a DNA sequence to RNA.

    Args:
        dna_sequence (str): The DNA sequence to be transcribed.

    Returns:
        str: The transcribed RNA sequence.
    """

    # Convert lowercase letters to uppercase
    dna_sequence = dna_sequence.upper()

    # Check for invalid characters
    for nucleotide in dna_sequence:
        if nucleotide not in valid_nucleotides:
            raise ValueError("Invalid DNA sequence.")

    # Replace T with U
    rna_sequence = dna_sequence.replace('T', 'U')
    return rna_sequence
```

# Task 2: Write a unit test
his task focuses on verifying the correctness and reliability of the DNA transcription function using unit tests. Different test cases are used to ensure that the function correctly transcribes standard DNA sequences, properly handles lowercase input, and raises an error when invalid nucleotide characters are provided.

```python
import unittest
```

**Inputs:**

```python
dna_sequence = "ATCGAT"
expected_rna_sequence = "AUCGAU"
invalid_dna_sequence = "AXCGAX"
valid_nucleotides = ['A', 'T', 'C', 'G']
```

**Unit test:**

```python

class MyFirstUniTTests(unittest.TestCase):
    
    def test_transcribe_dna_to_rna(self):
        """ Test standard DNA transcription. """

        self.assertEqual(transcribe_dna_to_rna(dna_sequence), expected_rna_sequence)

    def test_transcribe_dna_to_rna_lowercase(self):
        """ Test lowercase DNA transcription. """
        self.assertEqual(transcribe_dna_to_rna(dna_sequence.lower()), expected_rna_sequence)

    def test_transcribe_dna_to_rna_invalid_nucleotides(self):
        """ Test invalid DNA characters. """
        try:
          transcribe_dna_to_rna(invalid_dna_sequence)

        except ValueError:
          self.assertTrue(True)

        else:
          self.fail("ValueError not raised for invalid nucleotides")
```

Run the test

```python
if __name__ == '__main__':
    unittest.main(argv=['ignored', '-v'], exit=False)
```

**Comments on the result** : <br>
The result confirms that the function correctly handles lowercase DNA input by converting it into uppercase format before transcription. This improves the robustness and usability of the function.

# **Task 3: Random DNA Sequence Generation**
This task focuses on generating artificial (synthetic) DNA sequences for testing and experimentation purposes. The function randomly creates DNA strings using the valid nucleotide bases A, T, C, and G, with sequence lengths ranging between 10 and 100 characters. Such synthetic data can be useful for algorithm development, debugging, visualization, and data exploration in scientific programming.


```python
import random
```

```python
def generate_random_dna_sequence():
    """
    Randomly generates a single DNA sequence string.
    The sequence is composed of 'A', 'T', 'C', 'G'.
    The length is randomly chosen between 10 and 100.
    """

    sequence_length = random.randint(10,100)

    #Variable of the random sequence
    random_composed_sequence = ''

    for i in range(sequence_length):
        random_composed_sequence += random.choice(valid_nucleotides)
    return random_composed_sequence
```

```python
result = generate_random_dna_sequence()
print(result)
```

**Comments on the result:** <br> The generated output demonstrates that the function successfully creates synthetic DNA sequences using valid nucleotide bases only (A, T, C, and G). The sequence length is randomly generated, allowing the creation of diverse toy data for future testing and experimentation.
