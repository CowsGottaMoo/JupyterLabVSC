# Unit Test: DNA Transcription

**Goal:** The goal of this assignment is to practice unit testing with a small DNA transcription example and to generate simple toy DNA data for later algorithm development.

In this notebook a DNA sequence is transcribed to RNA by replacing thymine (`T`) with uracil (`U`). The transcription function is checked with unit tests before it is used on generated toy DNA sequences.

## Problem Description

DNA sequences are built from the four nucleotide bases:

- `A` for adenine
- `T` for thymine
- `C` for cytosine
- `G` for guanine

For transcription to RNA, every `T` is replaced by `U`. The valid input characters are therefore only `A`, `T`, `C`, and `G`. Lower-case input should also work by first converting the DNA sequence to upper case.

## Task 1: Transcription Function

First, the allowed libraries are imported. Then a user-defined function is written that validates a DNA string and transcribes it into RNA.

```python
# Importing the allowed libraries
import random
import unittest
```

```python
# Defining a function to transcribe DNA to RNA
def transcribe_dna_to_rna(dna_sequence: str) -> str:
    """Transcribe a DNA sequence to its RNA equivalent.
    
    Args:
        dna_sequence: DNA sequence containing the nucleotides A, T, C, and G.
    
    Returns:
        RNA sequence where each T has been replaced by U.
    """
    if not isinstance(dna_sequence, str):
        raise TypeError("dna_sequence must be a string")
    
    dna_sequence = dna_sequence.upper()
    allowed_nucleotides = ["A", "T", "C", "G"]
    
    for nucleotide in dna_sequence:
        if nucleotide not in allowed_nucleotides:
            raise ValueError("dna_sequence can only contain A, T, C, and G")
    
    rna_sequence = dna_sequence.replace("T", "U")
    return rna_sequence
```

Before writing the formal unit tests, the transcription function is tried once on the example sequence from the assignment.

```python
# Testing the transcription function with one simple example
example_dna_sequence = "ATCGAT"
example_rna_sequence = transcribe_dna_to_rna(dna_sequence=example_dna_sequence)

print(f"DNA sequence: {example_dna_sequence}")
print(f"RNA sequence: {example_rna_sequence}")
```

## Task 2: Unit Tests

The assignment asks for three checks:

1. a standard DNA sequence transcribes correctly,
2. lower-case input is handled correctly, and
3. a `ValueError` is raised when an invalid character is used.

The following test class uses the `unittest` library for these checks.

```python
# Defining the unit tests for the transcription function
class TestDNATranscription(unittest.TestCase):
    """Unit tests for the DNA transcription function."""
    
    def test_standard_sequence(self):
        """Check that a standard DNA sequence is transcribed correctly."""
        dna_sequence = "ATCGAT"
        expected_rna_sequence = "AUCGAU"
        observed_rna_sequence = transcribe_dna_to_rna(dna_sequence=dna_sequence)
        
        self.assertEqual(observed_rna_sequence, expected_rna_sequence)
    
    def test_case_insensitivity(self):
        """Check that lower-case input is converted and transcribed correctly."""
        dna_sequence = "atcg"
        expected_rna_sequence = "AUCG"
        observed_rna_sequence = transcribe_dna_to_rna(dna_sequence=dna_sequence)
        
        self.assertEqual(observed_rna_sequence, expected_rna_sequence)
    
    def test_invalid_character_raises_value_error(self):
        """Check that invalid DNA characters raise a ValueError."""
        value_error_raised = False
        
        try:
            transcribe_dna_to_rna(dna_sequence="ATXG")
        except ValueError:
            value_error_raised = True
        
        self.assertTrue(value_error_raised)


# Running the unit tests inside the notebook
test_suite = unittest.TestLoader().loadTestsFromTestCase(TestDNATranscription)
test_runner = unittest.TextTestRunner(verbosity=2)
test_result = test_runner.run(test_suite)

print()
print(f"All tests passed: {test_result.wasSuccessful()}")
assert test_result.wasSuccessful()
```

## Task 3: Toy Data

Next, a user-defined function is written that generates one random DNA sequence. The length is also selected randomly between a minimum and maximum length.

```python
# Defining a function to generate random DNA toy data
def generate_random_dna_sequence(minimum_length: int = 10, maximum_length: int = 100) -> str:
    """Generate one random DNA sequence.
    
    Args:
        minimum_length: Smallest possible sequence length.
        maximum_length: Largest possible sequence length.
    
    Returns:
        Random DNA sequence containing only A, T, C, and G.
    """
    if not isinstance(minimum_length, int):
        raise TypeError("minimum_length must be an integer")
    if not isinstance(maximum_length, int):
        raise TypeError("maximum_length must be an integer")
    if minimum_length < 1:
        raise ValueError("minimum_length must be at least 1")
    if maximum_length < minimum_length:
        raise ValueError("maximum_length must be greater than or equal to minimum_length")
    
    dna_nucleotides = ["A", "T", "C", "G"]
    sequence_length = random.randint(minimum_length, maximum_length)
    dna_sequence = ""
    
    for nucleotide_index in range(sequence_length):
        dna_sequence = dna_sequence + random.choice(dna_nucleotides)
    
    return dna_sequence
```

To make the notebook output reproducible after running it again, the random seed is set before generating example toy sequences. The function itself still uses random choices.

```python
# Generating example toy DNA sequences
random.seed(60)
number_of_sequences = 5

toy_dna_sequences = []

for sequence_index in range(number_of_sequences):
    toy_dna_sequence = generate_random_dna_sequence()
    toy_dna_sequences.append(toy_dna_sequence)

for sequence_index, toy_dna_sequence in enumerate(toy_dna_sequences, start=1):
    print(f"Toy DNA sequence {sequence_index}:")
    print(f"Length: {len(toy_dna_sequence)}")
    print(toy_dna_sequence)
    print()
```

Finally, the generated toy sequences are checked and transcribed. This confirms that the generated data is valid input for the transcription function.

```python
# Checking and transcribing the generated toy DNA sequences
allowed_nucleotides = ["A", "T", "C", "G"]

toy_rna_sequences = []

for toy_dna_sequence in toy_dna_sequences:
    assert 10 <= len(toy_dna_sequence) <= 100
    
    for nucleotide in toy_dna_sequence:
        assert nucleotide in allowed_nucleotides
    
    toy_rna_sequences.append(transcribe_dna_to_rna(dna_sequence=toy_dna_sequence))

print("All generated toy DNA sequences have valid lengths and valid nucleotides.")
print()

for sequence_index, toy_rna_sequence in enumerate(toy_rna_sequences, start=1):
    print(f"Toy RNA sequence {sequence_index}:")
    print(toy_rna_sequence)
    print()
```

## Comment on the Results

The unit tests verify that the transcription function works for a normal DNA sequence, handles lower-case input by converting it to upper case, and rejects corrupted input with invalid characters. This is important because invalid DNA input should not silently move into later analysis steps.

The toy data function creates random DNA sequences with lengths between 10 and 100 characters. The additional checks confirm that each generated sequence contains only valid DNA nucleotides and can therefore be transcribed with the same function.
